```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\BIT\Downloads\cleaned dataset for sql.csv")

df.head()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Customer Value Score",
    bins=20,
    kde=True
)

plt.title("Distribution of Customer Value Score")
plt.xlabel("Customer Value Score")
plt.ylabel("Number of Customers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Value Tier",
    order=df["Value Tier"].value_counts().index
)

plt.title("Customer Distribution by Value Tier")
plt.xlabel("Value Tier")
plt.ylabel("Number of Customers")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.boxplot(
    data=df,
    x="Value Tier",
    y="Purchase Amount (USD)"
)

plt.title("Purchase Amount by Customer Value Tier")
plt.xlabel("Value Tier")
plt.ylabel("Purchase Amount (USD)")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Purchase Amount (USD)",
    bins=20,
    kde=True
)

plt.title("Distribution of Customer Purchase Amount")
plt.xlabel("Purchase Amount (USD)")
plt.ylabel("Number of Customers")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Previous Purchases",
    bins=20,
    kde=True
)

plt.title("Distribution of Previous Purchases")
plt.xlabel("Number of Previous Purchases")
plt.ylabel("Number of Customers")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="Frequency of Purchases",
    order=df["Frequency of Purchases"].value_counts().index
)

plt.title("Customer Purchase Frequency")
plt.xlabel("Number of Customers")
plt.ylabel("Frequency of Purchases")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Subscription Status"
)

plt.title("Customer Subscription Status")
plt.xlabel("Subscription Status")
plt.ylabel("Number of Customers")

plt.show()
```

```python
numeric_columns = [
    "Age",
    "Purchase Amount (USD)",
    "Review Rating",
    "Previous Purchases",
    "Customer Value Score"
]

correlation_matrix = df[numeric_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Customer Metrics")

plt.show()
```
